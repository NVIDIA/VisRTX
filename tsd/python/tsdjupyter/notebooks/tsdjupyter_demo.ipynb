# TSD Jupyter — Remote rendering in Jupyter

This notebook shows how to use the **tsdjupyter** package to connect to a TSD render server and control the view and volume transfer functions from Jupyter.

**Prerequisites:** A TSD server must be running.

```python
from tsdjupyter import TSDJupyter, TransferFunctionWidget
```

```python
viewer = TSDJupyter(
    host="localhost",
    port=49000,
    width=1120,
    height=512,
    jpeg_quality=50,
    detachable=True,
)
tf_widget = TransferFunctionWidget(viewer)
```

Display the viewer and the transfer function widget. Use the canvas to orbit/dolly/pan; use the TF widget to edit colour and opacity for the selected volume.

```python
display(viewer)
```
