# TSD Viewer — Remote rendering in Jupyter

This notebook shows how to use the **tsdviewer** package to connect to a TSD render server and control the view and volume transfer functions from Jupyter.

**Prerequisites:** A TSD server must be running.

```python
from tsdviewer import TSDViewer, TransferFunctionWidget
```

```python
viewer = TSDViewer(
    host="localhost",
    port=12345,
    width=1120,
    height=256,
    jpeg_quality=50,
    detachable=True,
)
tf_widget = TransferFunctionWidget(viewer)
```
